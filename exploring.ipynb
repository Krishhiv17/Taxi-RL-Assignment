```python
import gymnasium as gym
import numpy as np

# Create env with text rendering
env = gym.make("Taxi-v3", render_mode="ansi")

# Helper: reset & return just the observation (state id)
def reset_env(env):
    obs, info = env.reset()
    return obs

# Helper: step and unify Gymnasium's (obs, reward, terminated, truncated, info)
def step_env(env, action):
    obs, reward, terminated, truncated, info = env.step(action)
    done = terminated or truncated
    return obs, reward, done, info

# Helper: print the grid
def render_env(env):
    print(env.render())
```

```python
action_meaning = {
    0: "South",
    1: "North",
    2: "East",
    3: "West",
    4: "Pickup",
    5: "Dropoff",
}

for a in range(env.action_space.n):
    print(a, "-", action_meaning[a])
```

```python
def demo_episode(actions, max_steps=20):
    state = reset_env(env)
    print("Initial state id:", state)
    render_env(env)

    total_reward = 0

    for step, action in enumerate(actions, start=1):
        print(f"\nStep {step}: action {action} ({action_meaning[action]})")
        next_state, reward, done, info = step_env(env, action)

        render_env(env)
        print("Reward:", reward)
        print("Next state id:", next_state)

        total_reward += reward

        if done:
            print("\nEpisode ended here.")
            break

    print("\nTotal reward this demo:", total_reward)
```

```python
demo_episode(actions=[0, 0, 0, 4, 0, 0, 5])  # totally arbitrary example
```
